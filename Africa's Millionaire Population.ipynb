Introduction

The distribution and growth of Africa’s millionaire population serve as key indicators of the continent’s economic performance, investment potential, and wealth mobility. Tracking changes in high-net-worth individuals provides insights into capital formation, market confidence, and regional competitiveness. Between 2015 and 2025, several African economies such as Mauritius, Morocco, and Rwanda have recorded strong millionaire growth, reflecting stable governance and expanding investment opportunities. In contrast, countries like Nigeria, Egypt, and Angola have experienced declines linked to currency pressures, political uncertainty, and capital flight. This analysis examines the trends and implications of Africa’s millionaire population to better understand emerging investment hubs, shifting wealth patterns, and their impact on long-term economic growth.

```python
#Visualise the dataframe:
df = pd.DataFrame(data)

# Horizontal bar chart for Millionaires
fig, ax = plt.subplots(figsize=(10, 6))
df_sorted = df.sort_values("Millionaires (USD $1M+)", ascending=True)
ax.barh(df_sorted["Country"], df_sorted["Millionaires (USD $1M+)"], color="teal")
ax.set_xlabel("Millionaires (USD $1M+)")
ax.set_title("Number of Millionaires by Country in Africa (2025)")
plt.tight_layout()

plt.show()
```

```python
# Visualise Data Based On the Percentage Growth
df = pd.DataFrame(data)

# Sort by Growth for better visualization
df_sorted = df.sort_values("Growth 2015‑2025 (%)", ascending=True)

# Horizontal bar chart for Growth (%)
fig, ax = plt.subplots(figsize=(10, 6))
colors = ["red" if x < 0 else "green" for x in df_sorted["Growth 2015‑2025 (%)"]]
ax.barh(df_sorted["Country"], df_sorted["Growth 2015‑2025 (%)"], color=colors)
ax.set_xlabel("Growth (%) from 2015 to 2025")
ax.set_title("African Millionaires Growth by Country (2015–2025)")
plt.tight_layout()

plt.show()
```
