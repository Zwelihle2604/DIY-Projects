Key events and their impact

COVID‑19 pandemic & lockdowns (2020)

In March 2020 South Africa declared a nationwide lockdown from 26 March onwards. 

The lockdown severely impacted business operations: in early April, nearly 90% of businesses reported below‑normal turnover; 48% temporarily ceased activity and 9% closed permanently. 

According to the South African Chamber of Commerce and Industry (SACCI) Business Confidence Index (BCI) for April 2020:

“The sharp deterioration … followed the worldwide Covid‑19 pandemic … The lockdown had major effect on economic activity and the business climate.” 


This event corresponds to one of the deepest drops in business confidence in recent years.

Persistent energy crisis / load‑shedding

South Africa has endured long‑running problems with electricity supply (“load‑shedding”), which has been estimated to reduce potential GDP growth by ~1–1.3% annually. 

Frequent power outages raise costs, disrupt manufacturing and service sectors, and undermine business sentiment.

July 2021 unrest & looting

In July 2021, nationwide unrest and looting (sparked partly by the imprisonment of a former president) caused major disruption. 

The event eroded confidence, especially in affected provinces and sectors dependent on infrastructure and logistics.

Global commodity price swings & trade dynamics

As a relatively open economy, South Africa is sensitive to export/import volumes, exchange‐rate swings, and global demand for commodities. The SACCI July 2021 report highlights improved export volumes in H1 2021, which bolstered confidence. 

Conversely, when exports/imports decline, or the rand weakens, business sentiment suffers.

Sectoral shocks and labour unrest

For example, the 2021 South African metalworkers strike (October 2021) disrupted manufacturing, which can feed into broader business confidence. 

Such strikes raise costs, interrupt supply chains, and raise uncertainty.

Delays or uncertainty in policy / budget / governance

Uncertainty about government policy, fiscal stance, and regulation tends to weigh on business confidence. For example, the unrest, budget disruptions, and weak investment sentiment all play a role. Conclusion 

Business confidence from 2017 to 2024 was highly influenced by major disruptions. The COVID-19 pandemic in 2020 caused the sharpest drop, while civil unrest and political instability in 2021 further dampened sentiment. Gradual recovery through 2022–2023 reflects adaptation and easing of restrictions, but challenges like energy shortages and policy uncertainty continued to constrain optimism. This data highlights how both external shocks and domestic instability directly affect business sentiment, emphasizing the need for resilient policies and infrastructure to maintain confidence.
